###  Word2Vec

- **Word → To → Vector**  
  Yani words ko numbers (vectors) me convert karna — *with meaning*.


### Word2Vec Ka Kaam

#### 1) Similar words ko close number vector me lana
- “king” and “queen” → near  
- “apple” and “mango” → near  
- “car” and “bus” → near  

#### 2) Relationship detect karna
- King → Queen  
- Man → Woman  
- Paris → France  
- Tokyo → Japan  


### Word2Vec Kaise Sikhta Hai?

- **CBOW (Continuous Bag of Words)** → **FIND OUT MISSING WORD**  
- **Skip-Gram** → **FIND OUT PREDICT WORD**

    - **Small dataset = CBOW**  
    - **Large dataset = Skip-Gram**

#### Improve:
- Increasing the training data  
- Increasing the window size  
- Vector size bhi increase hota hai  


### Dono Methods Detail

#### 1) CBOW  
Context (surrounding words) → Target word predict  
Example:  
“I ___ food”  
Predict: **love**

#### 2) Skip-Gram  
Target word → Context words predict  
Example:  
“love” → predict: **I, food**


### Why Word2Vec is Powerful?

- Words ka **meaning** capture karta hai  
- TF-IDF / BoW me meaning nahi hota  
- Semantic + relationships samajhta hai  
- Famous logic: **king − man + woman = queen**  


### Adavantegges

- sparse matrix ----------> dense matrix
- semantic info capture  
- vocablory size  ----------> fixe set of dimention (google dimention is [300 dimention])
- out os vocablory (oov) is also solved
